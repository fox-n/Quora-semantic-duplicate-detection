```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import paired_cosine_distances
```

```python
model = SentenceTransformer('/Users/nadiiababanska/Desktop/claude_coowork/ML_final_project/models/sentence_transformer_duplicate_model')  # локальний шлях в репо

pairs = [
    ("How do I learn Python?", "What is the best way to learn Python?"),      # очікуємо: дублікат
    ("How do I learn Python?", "What is the capital of France?"),              # очікуємо: не дублікат
    ("What is the best phone under 20000?", "Which is the best mobile phone under 20000 rupees?"),  # дублікат
]

for q1, q2 in pairs:
    emb1 = model.encode([q1])
    emb2 = model.encode([q2])
    sim = 1 - paired_cosine_distances(emb1, emb2)
    label = "DUPLICATE" if sim[0] > 0.60 else "not duplicate"  # твій best_threshold
    print(f"{sim[0]:.3f} -> {label}   |  {q1!r} vs {q2!r}")
```
